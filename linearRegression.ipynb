# Linear regression with SVD

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = 3
a = np.linspace(-2, 2, 100)
b = a * x + np.random.randn(*a.shape)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(a, a * x, label='y = 3x', linewidth=2, color='red')
plt.scatter(a, b, s=10, alpha=0.6, label='Noisy data')

plt.xlabel('a')
plt.ylabel('b')
plt.title('Noisy linear relationship')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```
